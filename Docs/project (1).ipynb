```python
train_df_nan = train_df.replace("?", np.nan)
test_df_nan = test_df.replace("?", np.nan)
train_df_nan.shape
```

```python
def clean_text(text):
    text = str (text).lower()
    text = re.sub(r'https?://\S+|www\.\S+','',text)
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    return text.strip()

```

```python
train_df_nan['clean_text'] = train_df_nan['text'].apply(clean_text)
test_df_nan['clean_text'] = test_df_nan['text'].apply(clean_text)
test_def_nan['clean_location'] = train_df_nan['location'].apply(clean_text)
test_def_nan['clean_location'] = test_df_nan['location'].apply(clean_text)
test_def_nan['clean_Keyword'] = test_df_nan['Keyword'].apply(clean_text)
test_def_nan['clean_Keyword'] = test_df_nan['Keyword'].apply(clean_text)
```
